```python
import json
import pandas as pd
import seaborn as sns

with open("../artifacts/results/cb-ret10-two-stage.json") as f:
    df = pd.DataFrame(json.load(f))

df["step"] += (df["stage"] == "relearning") * 32
df = df.melt(
    id_vars=["stage", "step"],
    value_vars=["wmdp_bio", "mmlu_no_bio"],
    var_name="metric",
    value_name="accuracy",
)

sns.lineplot(df, x="step", y="accuracy", hue="metric", style="stage", marker="o")
```
