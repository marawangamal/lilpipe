```python
import json
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

path = "/home/mila/m/marawan.gamal/projects/lilpipe/examples/weight-steering/artifacts/analysis/cosine-similarity-across-seeds/results.json"
cosim = np.array(json.load(open(path)))
labels = [
    f"{b}-{s}" for b in ("honesty", "non-syco", "non-cheat") for s in range(42, 47)
]

plt.figure(figsize=(10, 8))
sns.heatmap(
    cosim,
    mask=np.eye(len(cosim), dtype=bool),
    xticklabels=labels,
    yticklabels=labels,
    square=True,
)
plt.xticks(rotation=55, ha="right")
plt.tight_layout()
plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

path = "/home/mila/m/marawan.gamal/projects/lilpipe/examples/weight-steering/artifacts/analysis/conic-distance-across-seeds/results.json"
cone_dist = np.array(json.load(open(path))).T

seed_labels = [
    f"{behavior}-{seed}"
    for behavior in ("honesty", "non-syco", "non-cheat")
    for seed in range(42, 47)
]
cone_labels = ["honesty cone", "non-syco cone", "non-cheat cone"]

plt.figure(figsize=(15, 3))
sns.heatmap(
    cone_dist,
    xticklabels=seed_labels,
    yticklabels=cone_labels,
    vmin=0,
    vmax=1,
    cmap="viridis",
    annot=True,
    fmt=".2f",
)
plt.xticks(rotation=55, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
