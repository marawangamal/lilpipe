```python
import json
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

path = "/home/mila/m/marawan.gamal/projects/lilpipe/examples/weight-steering/artifacts/analysis/cosine-similarity-across-seeds/results.json"
cosim = np.array(json.load(open(path)))
labels = [
    f"{b}-{s}" for b in ("honesty", "non-syco", "non-cheat") for s in range(42, 47)
]

plt.figure(figsize=(10, 8))
sns.heatmap(
    cosim,
    mask=np.eye(len(cosim), dtype=bool),
    xticklabels=labels,
    yticklabels=labels,
    square=True,
)
plt.xticks(rotation=55, ha="right")
plt.tight_layout()
plt.show()
```
